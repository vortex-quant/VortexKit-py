```python
df_precision = agg.detect_timestamp_precision(df["timestamp"])
print(f"\nDetected Timestamp Precision: {df_precision}")
```

```python
klines_5m = agg.aggregate_klines(df, interval="5m",precision=df_precision)
klines_1h = agg.aggregate_klines(df, interval="1h",precision=df_precision)

print(f"5m klines: {klines_5m.shape}")
print(f"1h klines: {klines_1h.shape}")
```
